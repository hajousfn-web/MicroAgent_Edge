# MicroAgent-Edge submission archive builder

This notebook clones the public MicroAgent-Edge repository, runs its `build_submission.py` verification, and places the resulting `submission.zip` in `/kaggle/working/` for download. The kernel needs internet access to clone the source repository. The attached competition source provides the Gemma 4 competition context; this archive-building notebook does not run benchmark inference or certify that the archive meets the official submission schema.

```python
from pathlib import Path
import shutil
import subprocess
import sys
import tempfile

repo_url = "https://github.com/hajousfn-web/MicroAgent_Edge.git"
output_zip = Path("/kaggle/working/submission.zip")

with tempfile.TemporaryDirectory(prefix="microagent-edge-") as workspace:
    repo_dir = Path(workspace) / "MicroAgent_Edge"
    subprocess.run(
        ["git", "clone", "--depth", "1", repo_url, str(repo_dir)],
        check=True,
    )

    subprocess.run(
        [sys.executable, "build_submission.py"],
        cwd=repo_dir,
        check=True,
    )

    built_zip = repo_dir / "submission.zip"
    if not built_zip.is_file():
        raise FileNotFoundError(f"Archive builder did not create {built_zip}")
    shutil.copy2(built_zip, output_zip)
print(f"Verified submission archive: {output_zip} ({output_zip.stat().st_size:,} bytes)")
```

When the cell completes, use the notebook output panel to download `/kaggle/working/submission.zip`. The archive is a packaging smoke-test, not a validated competition submission.
